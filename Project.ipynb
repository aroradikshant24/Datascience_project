```python
''' NaN — Not a Number, commonly used for missing numeric data.
None — Python's absence-of-value object.
pd.NA — Pandas' nullable missing-value marker.
NaT — Missing datetime value.                                   '''
# Note - an empty string "" or a string containing only spaces "   " is not automatically considered missing by isna() when already stored in a DataFrame.

''' Scratch logic for the above method '''

# missing = df.isna()
# print(missing.head())

##to learn
import math

def is_missing(value):
    if value is None:
        return True

    try:
        if value != value:
            return True
    except (TypeError, ValueError):
        pass

    if isinstance(value, float):
        return math.isnan(value)

    return False

print(is_missing(None))
print(is_missing(float("nan")))
print(is_missing(100))

```

```python
# 1. For each column, counting missing values.
#missing_count = df.isna().sum()
#print(missing_count)

''' Scratch logic for the above method '''

missing_count_manual = {}

for col in df.columns:
    count = 0
    for value in df[col]:
        if is_missing(value):
            count += 1
    missing_count_manual[col] = count

print(missing_count_manual)

# 2. for whole dataset, counting missing values
# total_missing = df.isna().sum().sum()
# print(total_missing)

''' Scratch logic for the above method '''


total_manual = 0

for col in df.columns:
    for value in df[col]:
        if is_missing(value):
            total_manual += 1

print(total_manual)

'''df.isna().sum(axis=0) #for each column
   df.isna().sum(axis=1) #for each row   '''

```

```python
''' Missing-value treatment is also called imputation when we estimate and substitute a missing value.
    There is no single correct method for every column.
    Method: Suitable use
    1. Mean imputation: Approximately symmetric numerical distributions
    2. Median imputation: Skewed numerical distributions or outliers
    3. Mode imputation:	Categorical columns, when justified
    4. Business-formula: reconstruction	Values determined by other reliable fields
    5. Leave missing:	Information that cannot be reliably inferred
    6. Delete records:	When justified by the amount and pattern of missingness         '''

# 1. Mean Imputation on Shipping Cost field

# col = "Shipping Cost (₹)"

# mean_shipping = df_clean[col].mean()
# shipping_builtin = (df_clean[col].fillna(mean_shipping))

# print(mean_shipping)

"Scratch logic for above mean imputation"

col = "Shipping Cost (₹)"

total = 0
count = 0

for value in df_clean[col]:
    if not pd.isna(value):
        total = total + value
        count = count + 1

mean = total / count

for i in range(len(df_clean)):
    if pd.isna(df_clean.loc[i, col]):
        df_clean.loc[i, col] = mean

print("Mean Shipping Cost:", mean)

# first 10 rows after performing mean imputation
display(df_clean.head(10))

```

```python
# 2. Median Imputation on Unit Cost field

# col = "Unit Cost (₹)"

# median_value = df_clean[col].median()
# unit_cost_builtin = (df_clean[col].fillna(median_value))

# print(median_value)

"Scratch logic for above median imputation"

col="Unit Cost (₹)"
values = []

for value in df_clean[col]:
    if pd.notna(value):
        values.append(value)

values.sort()
n=len(values)

if n % 2 == 0:
    median = (values[n//2 - 1] + values[n//2]) / 2
else:
    median = values[n//2]

print("Median Unit Cost:", median)

for i in range(len(df_clean)):
    if pd.isna(df_clean.loc[i, col]):
        df_clean.loc[i, col] = median

print(df_clean[col])

```

```python
# 3. Mode Imputation on Category Field

# col = "Category"

# mode_value = (df_clean[col].mode().iloc[0])
# category_builtin = (df_clean[col].fillna(mode_value))

# print(mode_value)

''' scratch logic for above mode imputation '''

frequency = {}

for value in df_clean["Category"]:
    if pd.notna(value):
        if value in frequency:
            frequency[value] += 1
        else:
            frequency[value] = 1

highest = max(frequency.values())

for key in frequency:
    if frequency[key] == highest:
        print(key)

for i in range(len(df_clean)):
    if pd.isna(df_clean.loc[i, "Category"]):
        df_clean.loc[i, "Category"] = key

print(df_clean["Category"])
```

```python
# 4. Business formula imputation on Taxable value field

# qty = df_clean["Quantity (Units)"]
# cost = df_clean["Unit Cost (₹)"]

# taxable_calculated = qty * cost

# taxable_filled = (df_clean["Taxable Value (₹)"].fillna(taxable_calculated))

''' Scratch logic for above business formula imputation '''

for i in range(len(df_clean)):
    if pd.isna(df_clean.loc[i, "Taxable Value (₹)"]):
      qty = df_clean.loc[i, "Quantity (Units)"]
      cost = df_clean.loc[i, "Unit Cost (₹)"]

      if pd.notna(qty) and pd.notna(cost):
        df_clean.loc[i, "Taxable Value (₹)"] = qty * cost

print(df_clean["Taxable Value (₹)"])

```

```python
#Applying IQR method for outlier detection
''' col = "Unit Cost (₹)"

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    outliers = df_clean[ (df_clean[col] < lower) | (df_clean[col] > upper) ]

    Extract the rows containing potential outliers and then storing in a new dataframe.

    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower:", lower)
    print("Upper:", upper)
    print("Outliers:", len(outliers)) '''

# Any Unit Cost greater than ₹8,850 is flagged as a potential outlier.
# The negative lower bound is only a statistical threshold; it does not mean negative unit costs are valid. '''

''' Scratch logic for above method '''

values = []

for value in df_clean["Unit Cost (₹)"]:
    if pd.notna(value):
      values.append(value)

values.sort()
n = len(values)

q1 = values[int(n * 0.25)]
q3 = values[int(n * 0.75)]

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = []

for value in values:
    if value < lower or value > upper:
      outliers.append(value)

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Limit:", lower)
print("Upper Limit:", upper)
print("Outliers:", outliers)
print("Total Outliers:", len(outliers))
```
